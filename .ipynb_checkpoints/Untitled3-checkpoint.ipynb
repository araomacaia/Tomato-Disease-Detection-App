```python
import tensorflow as tf
import os

# Path to your saved model directory (the one ending with "1")
old_model_path = "tomato_disease_model/1"

# Check if the path exists
if not os.path.exists(old_model_path):
    raise FileNotFoundError(f"Model path not found: {old_model_path}")

# Load the saved model (without using keras.models.load_model)
loaded_model = tf.saved_model.load(old_model_path)

# Get the inference function
infer = loaded_model.signatures['serving_default']

# Convert to TFLite using the SavedModel format directly
converter = tf.lite.TFLiteConverter.from_saved_model(old_model_path)
tflite_model = converter.convert()

# Save the new TFLite file
tflite_path = "tomato_disease_model/tomato_model.tflite"
with open(tflite_path, "wb") as f:
    f.write(tflite_model)

print(f"TFLite model successfully converted and saved to: {tflite_path}")
```
