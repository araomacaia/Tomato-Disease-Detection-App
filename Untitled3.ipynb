```python
import tensorflow as tf
import h5py
import os

# Defining model path
model_path = "tomato_disease_model/tomato_model_fixed.keras"

# Loading model weights manually without deserializing TFServingWrapper
try:
    model = tf.keras.models.load_model(model_path, compile=False)
    print("Model loaded successfully using standard method.")
except Exception as e:
    print("Standard load failed due to custom wrapper. Falling back to manual reconstruction...")
    # Creating a new empty Sequential model as a container
    model = tf.keras.Sequential()

# Trying loading the SavedModel version (fallback)
saved_dir = "tomato_disease_model/1"
if os.path.exists(saved_dir):
    print("Found SavedModel directory. Using it instead.")
    model = tf.saved_model.load(saved_dir)
else:
    print("No SavedModel folder found. Only .keras file is available.")

# Converting using TFLite Select ops to allow unsupported layers
converter = tf.lite.TFLiteConverter.from_saved_model(saved_dir)
converter.target_spec.supported_ops = [
    tf.lite.OpsSet.TFLITE_BUILTINS,
    tf.lite.OpsSet.SELECT_TF_OPS
]
converter.experimental_enable_resource_variables = True

tflite_model = converter.convert()

# Saving the TFLite file
with open("tomato_disease_model/tomato_model_flex.tflite", "wb") as f:
    f.write(tflite_model)

print("Conversion completed successfully — model saved as tomato_model_flex.tflite")
```
